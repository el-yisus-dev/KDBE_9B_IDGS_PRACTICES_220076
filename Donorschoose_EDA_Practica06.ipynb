 
# Análisis en profundidad y visualizaciones: DonorsChoose   

DonorsChoose.org es una organización sin fines de lucro con sede en Estados Unidos que permite a las personas donar directamente a proyectos de aula de escuelas públicas. Fundada en 2000 por Charles Best, entonces maestro de una escuela pública, DonorsChoose.org fue una de las primeras plataformas de financiamiento colectivo cívico de su tipo. La organización ha recibido la calificación más alta de Charity Navigator cada año desde 2005. En enero de 2018, anunciaron que se habían financiado 1 millón de proyectos. En el 77% de las escuelas públicas de Estados Unidos, al menos un proyecto ha sido solicitado en DonorsChoose.org. Las escuelas de zonas más acomodadas tienen más probabilidades de pedir tecnología, mientras que las escuelas de zonas menos favorecidas tienen más probabilidades de solicitar insumos básicos. También se ha observado que los donantes recurrentes en DonorsChoose normalmente donan a proyectos con los que no tienen una relación previa, y con mayor frecuencia financian proyectos que atienden a estudiantes con dificultades económicas.


## Ciencia de Datos para el Bien Común: DonorsChoose

DonorsChoose.org ha financiado más de 1.1 millones de solicitudes de aula gracias al apoyo de 3 millones de donantes, la mayoría de los cuales realizó su primera donación a una escuela pública. Si DonorsChoose.org puede motivar siquiera a una fracción de esos donantes a hacer otra donación, eso podría tener un gran impacto en la cantidad de solicitudes de aula que se cumplen.

Una buena solución permitirá a DonorsChoose.org crear campañas de correo electrónico dirigidas que recomienden solicitudes específicas de aula a donantes anteriores. Parte del reto es evaluar las necesidades de la organización, descubrir hallazgos a partir de los datos disponibles y construir la solución adecuada para este problema.

```python
def bar_hor(df, col, title, color, w=None, h=None, lm=0, limit=100, return_trace=False, rev=False):
    cnt_srs = df[col].value_counts()
    yy = cnt_srs.head(limit).index[::-1]
    xx = cnt_srs.head(limit).values[::-1]
    if rev:
        yy = cnt_srs.tail(limit).index[::-1]
        xx = cnt_srs.tail(limit).values[::-1]


    trace = go.Bar(y=yy, x=xx, orientation = 'h', marker=dict(color=color))
    if return_trace:
        return trace
    layout = dict(title=title, margin=dict(l=lm), width=w, height=h)
    data = [trace]
    fig = go.Figure(data=data, layout=layout)
    iplot(fig)

def bar_hor_noagg(x, y, title, color, w=None, h=None, lm=0, limit=100, rt=False):
    trace = go.Bar(y=x, x=y, orientation = 'h', marker=dict(color=color))
    if rt:
        return trace
    layout = dict(title=title, margin=dict(l=lm), width=w, height=h)
    data = [trace]
    fig = go.Figure(data=data, layout=layout)
    iplot(fig)


def bar_ver_noagg(x, y, title, color, w=None, h=None, lm=0, rt = False):
    trace = go.Bar(y=y, x=x, marker=dict(color=color))
    if rt:
        return trace
    layout = dict(title=title, margin=dict(l=lm), width=w, height=h)
    data = [trace]
    fig = go.Figure(data=data, layout=layout)
    iplot(fig)


trace1 = bar_hor(donors_df, 'Donor City', "Ciudades con más donantes", '#c2d2ed', 600, 400, 200, limit = 10, return_trace=True)
trace2 = bar_hor(donors_df, 'Donor State', "Estados con más donantes", '#c2d2ed', 600, 400, 200, limit = 10, return_trace=True)
# trace3 = bar_hor(donors_df, 'Donor City', "Ciudades con más donantes", '#4fe4f9', 600, 400, 200, limit = 10, return_trace=True, rev=True)
trace4 = bar_hor(donors_df, 'Donor State', "Estados con más donantes", '#c2d2ed', 600, 400, 200, limit = 10, return_trace=True, rev=True)


states_df = donors_df['Donor State'].value_counts()
statesdf = pd.DataFrame()
statesdf['state'] = states_df.index
statesdf['counts'] = states_df.values


scl = [[0.0, 'rgb(242,240,247)'],[0.2, 'rgb(218,218,235)'],[0.4, 'rgb(188,189,220)'], [0.6, 'rgb(158,154,200)'],[0.8, 'rgb(117,107,177)'],[1.0, 'rgb(84,39,143)']]
state_to_code = {'District of Columbia' : 'dc','Mississippi': 'MS', 'Oklahoma': 'OK', 'Delaware': 'DE', 'Minnesota': 'MN', 'Illinois': 'IL', 'Arkansas': 'AR', 'New Mexico': 'NM', 'Indiana': 'IN', 'Maryland': 'MD', 'Louisiana': 'LA', 'Idaho': 'ID', 'Wyoming': 'WY', 'Tennessee': 'TN', 'Arizona': 'AZ', 'Iowa': 'IA', 'Michigan': 'MI', 'Kansas': 'KS', 'Utah': 'UT', 'Virginia': 'VA', 'Oregon': 'OR', 'Connecticut': 'CT', 'Montana': 'MT', 'California': 'CA', 'Massachusetts': 'MA', 'West Virginia': 'WV', 'South Carolina': 'SC', 'New Hampshire': 'NH', 'Wisconsin': 'WI', 'Vermont': 'VT', 'Georgia': 'GA', 'North Dakota': 'ND', 'Pennsylvania': 'PA', 'Florida': 'FL', 'Alaska': 'AK', 'Kentucky': 'KY', 'Hawaii': 'HI', 'Nebraska': 'NE', 'Missouri': 'MO', 'Ohio': 'OH', 'Alabama': 'AL', 'Rhode Island': 'RI', 'South Dakota': 'SD', 'Colorado': 'CO', 'New Jersey': 'NJ', 'Washington': 'WA', 'North Carolina': 'NC', 'New York': 'NY', 'Texas': 'TX', 'Nevada': 'NV', 'Maine': 'ME'}
statesdf['state_code'] = statesdf['state'].apply(lambda x : state_to_code[x] if x in state_to_code else "")

data = [ dict(
        type='choropleth',
        colorscale = scl,
        autocolorscale = False,
        locations = statesdf['state_code'],
        z = statesdf['counts'],
        locationmode = 'USA-states',
        text = statesdf['state'],
        marker = dict(
            line = dict (
                color = 'rgb(255,255,255)',
                width = 2
            ) ),
        colorbar = dict(
            title = "Donantes")
        ) ]

layout = dict(
        title = 'Estados y el conteo de donantes',
        geo = dict(
            scope='usa',
            projection=dict( type='albers usa' ),
            showlakes = True,
            lakecolor = 'rgb(255, 255, 255)'),
             )

fig = dict( data=data, layout=layout )
iplot( fig, filename='d3-cloropleth-map' )
```

## 2.4 Donantes maestros y no maestros en los estados

```python
# which city and state has maximum teacher donors
tempdf = donors_df.groupby(['Donor State', 'Donor Is Teacher']).agg({'Donor Is Teacher' : 'count' }).rename(columns={'Donor Is Teacher' : 'Teacher Donor Counts'}).reset_index()

no_df = tempdf[tempdf['Donor Is Teacher']=='No']
no_x = no_df['Donor State']
no_y = no_df['Teacher Donor Counts']
yes_df = tempdf[tempdf['Donor Is Teacher']=='Yes']
yes_x = yes_df['Donor State']
yes_y = yes_df['Teacher Donor Counts']

trace1 = go.Bar(
    x=no_x,
    y=no_y,
    name='Non Teacher Donors',
    marker=dict(color="#751764"),
    opacity=0.8
)
trace2 = go.Bar(
    x=yes_x,
    y=yes_y,
    name='Teacher Donors',
    marker=dict(color="#b37f05"),
    opacity=0.8
)

data = [trace1, trace2]
layout = go.Layout(
    barmode='group',
    legend=dict(dict(x=-.1, y=1.2)),
    title = 'States and the distribution of Teacher and Non Teacher Donors',
)

fig = go.Figure(data=data, layout=layout)
iplot(fig, filename='grouped-bar')
```

```python
# ============================================================
# Coordenadas reales de ciudades de EE.UU. + funcion reutilizable
# Reemplaza el archivo zip_codes_states.csv del notebook original de Kaggle.
# ============================================================
c = pd.read_csv(
    "https://raw.githubusercontent.com/kelvins/US-Cities-Database/main/csv/us_cities.csv"
).rename(columns={'CITY': 'city', 'STATE_NAME': 'state',
                  'LATITUDE': 'latitude', 'LONGITUDE': 'longitude'})

# Una coordenada por (estado, ciudad)
coords = c.groupby(['state', 'city']).agg(
    {'latitude': 'mean', 'longitude': 'mean'}).reset_index()


def mapa_calor_estado(nombre_estado, centro, zoom=6, radius=12):
    """Mapa de calor de donantes por ciudad para un estado (datos reales)."""
    t = donors_df[donors_df['Donor State'] == nombre_estado]['Donor City'].value_counts()
    dons = pd.DataFrame({'city': t.index, 'count': t.values})
    cc = coords[coords['state'] == nombre_estado][['city', 'latitude', 'longitude']]
    datos = dons.merge(cc, on='city', how='left').dropna()

    mapa = folium.Map(centro, zoom_start=zoom, tiles='cartodbdark_matter')
    calor = [[r['latitude'], r['longitude'], r['count']] for _, r in datos.iterrows()]
    mapa.add_child(plugins.HeatMap(calor, radius=radius))
    return mapa


# 2.5.1 Donantes de California
mapa_calor_estado('California', [36.7783, -119.4179], zoom=6, radius=12)
```

> - Alta densidad alrededor de áreas metropolitanas como San Franscisco, Los Angles, San Jose y San Diago

## <a id="8.1">8.1 Limpieza de texto </a>

Se debe crear una función de utilidad que normalice el texto y elimine el ruido. Esta función de limpieza elimina las palabras vacías (*stopwords*) y los signos de puntuación del texto.

```python
from nltk.corpus import stopwords
import string

# Lista de stopwords
stop_words = set(stopwords.words("english"))

# Palabras de ruido adicionales
noise = {
    "students",
    "my",
    "title",
    "would",
    "every",
    "will",
    "many",
    "also",
    "donotremoveessaydivider"
}

def clean_text(txt):
    # Manejar valores nulos
    if not isinstance(txt, str):
        return ""

    # Minúsculas
    txt = txt.lower()

    # Eliminar puntuación
    txt = txt.translate(str.maketrans("", "", string.punctuation))

    # Eliminar stopwords
    words = [
        word
        for word in txt.split()
        if word not in stop_words
    ]

    clean_txt = " ".join(words)

    # Eliminar palabras de ruido
    for word in noise:
        clean_txt = clean_txt.replace(word, "")

    return " ".join(clean_txt.split())


# Hacer merge solo si aún no existe la columna
if "School State" not in projects_df.columns:
    projects_df = projects_df.merge(
        schools_df,
        on="School ID",
        how="inner"
    )

# Filtrar abril de 2018
april18_df = projects_df[
    (projects_df["Posted Year"] == 2018) &
    (projects_df["Posted Month"] == 4)
].copy()

# Limpiar texto
april18_df["clean_essay"] = april18_df["Project Essay"].apply(clean_text)
april18_df["clean_need"] = april18_df["Project Need Statement"].apply(clean_text)
```

```python
import pandas as pd

# Cargar proyectos
projects_df = pd.read_csv(
    "projects.csv"
)

# Cargar profesores
teachers_df = pd.read_csv(
    "teachers.csv"
)

# Cargar escuelas
schools_df = pd.read_csv(
    "schools.csv"
)

# Cargar donaciones
donations_df = pd.read_csv(
    "donations.csv"
)


projects_df = pd.read_csv(
    "projects.csv"
)


# Crear columnas de fecha
projects_df['Project Posted Date'] = pd.to_datetime(
    projects_df['Project Posted Date']
)


projects_df['Posted Year'] = (
    projects_df['Project Posted Date']
    .dt.year
)


projects_df['Posted Month'] = (
    projects_df['Project Posted Date']
    .dt.month
)
# ============================================
# Preparar datos
# ============================================

num_donations = (
    donations_df
    .groupby('Project ID')
    .agg({'Donation ID': 'count'})
    .reset_index()
    .rename(columns={'Donation ID': 'NumDonations'})
)

tempdf = projects_df.merge(num_donations, on="Project ID")


# Proyectos con un solo donador
df1 = tempdf[
    (tempdf['Posted Year'] == 2017) &
    (tempdf['NumDonations'] == 1)
].copy()


# Proyectos con múltiples donadores
df2 = tempdf[
    (tempdf['Posted Year'] == 2017) &
    (tempdf['NumDonations'] > 10)
].copy()


# ============================================
# Agregar información de profesores
# ============================================

df1 = df1.merge(
    teachers_df,
    on="Teacher ID",
    how="left"
)

df2 = df2.merge(
    teachers_df,
    on="Teacher ID",
    how="left"
)


# ============================================
# Agregar información de escuelas
# ============================================

# limpiar nombres de columnas
schools_df.columns = schools_df.columns.str.strip()

# corregir posible error del CSV
if 'chool ID' in schools_df.columns:
    schools_df.rename(
        columns={'chool ID': 'School ID'},
        inplace=True
    )


df1 = df1.merge(
    schools_df,
    on="School ID",
    how="left"
)

df2 = df2.merge(
    schools_df,
    on="School ID",
    how="left"
)


# ============================================
# Función para gráficas
# ============================================

def get_trace(col, func, flag=None, leg=False):

    t1 = df1[col].value_counts()
    t2 = df2[col].value_counts()

    if flag == 'num':
        t1 = df1[df1['NumProjects'] <= 20][col].value_counts()
        t2 = df2[df2['NumProjects'] <= 20][col].value_counts()


    trace1 = func(
        x=t1.index,
        y=[float(a)*100/sum(t1.values) for a in t1.values],
        name='Single Donors',
        marker=dict(color='#0de286'),
        showlegend=leg,
        opacity=0.9
    )


    trace2 = func(
        x=t2.index,
        y=[float(a)*100/sum(t2.values) for a in t2.values],
        name='Multiple Donors',
        marker=dict(color='#e82041'),
        opacity=0.9,
        showlegend=leg
    )


    return trace1, trace2



# ============================================
# Crear trazas
# ============================================

tr1, tr2 = get_trace(
    'Project Grade Level Category',
    go.Scatter
)

tr3, tr4 = get_trace(
    'School Metro Type',
    go.Scatter
)

tr5, tr6 = get_trace(
    'School State',
    go.Bar
)

tr7, tr8 = get_trace(
    'Project Resource Category',
    go.Bar
)

tr9, tr10 = get_trace(
    'Posted Month',
    go.Bar,
    leg=True
)


# ============================================
# Word Count
# ============================================

df1['wc'] = df1['Project Essay'].apply(lambda x: len(x.split()))
df2['wc'] = df2['Project Essay'].apply(lambda x: len(x.split()))


bins = [
    0, 50, 100, 150, 200,
    250, 300, 350,
    400, 450, 500, 10000
]


df1['wc'] = pd.cut(df1['wc'], bins)

t1 = dict(df1['wc'].value_counts())

labels = [
    str(b)
    .replace("]", "")
    .replace("(", "")
    .replace(", ", "-")
    for b in t1.keys()
]


vals = [
    float(a)*100/sum(t1.values())
    for a in list(t1.values())
]


d = {}

for i, v in enumerate(labels):
    d[v] = vals[i]


order = [
    '0-50',
    '50-100',
    '100-150',
    '150-200',
    '200-250',
    '250-300',
    '300-350',
    '350-400',
    '400-450',
    '450-500',
    '500-10000'
]


values = [d[v] for v in order]


tr13 = go.Bar(
    x=order,
    y=values,
    name='Single Donors',
    marker=dict(color='#0de286'),
    showlegend=False,
    opacity=0.9
)



df2['wc'] = pd.cut(df2['wc'], bins)

t2 = dict(df2['wc'].value_counts())

vals = [
    float(a)*100/sum(t2.values())
    for a in list(t2.values())
]

d = {}

for i, v in enumerate(labels):
    d[v] = vals[i]


values = [d[v] for v in order]


tr14 = go.Bar(
    x=order,
    y=values,
    name='Multiple Donors',
    marker=dict(color='#e82041'),
    showlegend=False,
    opacity=0.9
)



# ============================================
# Cost
# ============================================

bins = [
    0,
    100,
    500,
    1000,
    1500,
    2000,
    2500,
    3000,
    70000
]


df1['binnedcost'] = pd.cut(df1['cost'], bins)

t1 = dict(df1['binnedcost'].value_counts())

labels = [
    str(b)
    .replace("]", "")
    .replace("(", "")
    .replace(", ", "-")
    for b in t1.keys()
]


vals = [
    float(a)*100/sum(t1.values())
    for a in list(t1.values())
]


d = {}

for i, v in enumerate(labels):
    d[v] = vals[i]


order_cost = [
    '100-500',
    '500-1000',
    '1000-1500',
    '1500-2000',
    '2000-2500',
    '2500-3000',
    '3000-70000'
]


values = [d[v] for v in order_cost]


tr15 = go.Scatter(
    x=order_cost,
    y=values,
    name='Single Donors',
    fill='tozeroy',
    marker=dict(color='#0de286'),
    showlegend=False,
    opacity=0.9
)



df2['binnedcost'] = pd.cut(df2['cost'], bins)

t2 = dict(df2['binnedcost'].value_counts())


vals = [
    float(a)*100/sum(t2.values())
    for a in list(t2.values())
]


d = {}

for i, v in enumerate(labels):
    d[v] = vals[i]


values = [d[v] for v in order_cost]


tr16 = go.Scatter(
    x=order_cost,
    y=values,
    name='Multiple Donors',
    fill='tozeroy',
    marker=dict(color='#e82041'),
    showlegend=False,
    opacity=0.9
)



# ============================================
# Crear dashboard
# ============================================

titles = [
    'Project Grade Level Category',
    'School Metro Type',
    'Project Cost (in $)',
    'Project Essay Word Count',
    'Project Posted Month',
    'Resource Category'
]


fig = tools.make_subplots(
    rows=3,
    cols=2,
    subplot_titles=titles,
    print_grid=False,
    specs=[
        [{}, {}],
        [{}, {}],
        [{}, {}]
    ]
)


fig.append_trace(tr1, 1, 1)
fig.append_trace(tr2, 1, 1)

fig.append_trace(tr3, 1, 2)
fig.append_trace(tr4, 1, 2)

fig.append_trace(tr15, 2, 1)
fig.append_trace(tr16, 2, 1)

fig.append_trace(tr13, 2, 2)
fig.append_trace(tr14, 2, 2)

fig.append_trace(tr7, 3, 2)
fig.append_trace(tr8, 3, 2)

fig.append_trace(tr9, 3, 1)
fig.append_trace(tr10, 3, 1)


fig['layout'].update(
    barmode='group',
    title='Key Project Features Comparison: <b>Projects with Single Vs Multiple Donors</b><br>',
    titlefont=dict(
        size=22,
        color='#000'
    ),
    legend=dict(
        x=0.25,
        y=1.06,
        orientation="h"
    ),
    margin=dict(
        t=100,
        b=100
    ),
    paper_bgcolor='rgb(254, 247, 234)',
    plot_bgcolor='rgb(254, 247, 234)',
    height=1000
)


iplot(fig)
```
